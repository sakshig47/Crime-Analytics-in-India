```python
import pandas as pd

# Replace 'Null' with missing values
df["date_case_closed"] = df["date_case_closed"].replace("Null", pd.NA)

# Extract only the date (works only for non-null values)
df["date_case_closed"] = df["date_case_closed"].str.split().str[0]

# Convert to datetime
df["date_case_closed"] = pd.to_datetime(
    df["date_case_closed"],
    format="%d-%m-%Y",
    errors="coerce"
)
```
