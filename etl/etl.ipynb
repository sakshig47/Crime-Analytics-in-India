```python
# Create a new column based on time of occurrence
def crime_period(time):
    hour = time.hour

    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df["crime_occurred"] = df["time_of_occurrence"].apply(crime_period)
```

```python
df
```

```python
# Function to categorize age
def age_group(age):
    if age <= 12:
        return "Child"
    elif age <= 19:
        return "Teenager"
    elif age <= 59:
        return "Adult"
    else:
        return "Senior-Citizen"

# Create new column
df["age_group"] = df["victim_age"].apply(age_group)
```
